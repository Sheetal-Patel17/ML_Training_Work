Question 1: What is NumPy?
Definition

NumPy (Numerical Python) is a Python library used for fast numerical computing. It provides a powerful N-dimensional array (ndarray) and many built-in mathematical functions.

It is one of the most important libraries in Python and is widely used in:

Data Science
Machine Learning
Artificial Intelligence
Deep Learning
Scientific Computing
Image Processing
Data Analysis
Why do we use NumPy?

Python lists are good for storing data, but they become slow when working with large datasets.

NumPy solves this by:

Performing calculations much faster.
Using less memory.
Supporting multidimensional arrays.
Providing optimized mathematical functions.

```python
import numpy as np

marks = np.array([75, 80, 92, 68, 88])

print("Marks:", marks)
print("Average:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
```

Question 2. Why is NumPy faster than Python Lists?
First Understand Python List

Python List

[10,20,30,40]

Each element is stored separately in memory.

Different data types can exist together.

[10,"Hello",3.5,True]

This flexibility makes lists slower.

NumPy Array
[10 20 30 40]

Stores all values in one continuous block of memory.

All elements have the same data type.

This allows the CPU to process them much faster.

```python
import numpy as np
import time

# Python List
lst = list(range(1000000))

start = time.time()

lst = [x*2 for x in lst]

print("Python List Time:", time.time()-start)

# NumPy Array

arr = np.arange(1000000)

start = time.time()

arr = arr*2

print("NumPy Time:", time.time()-start)
```

Question 16. Broadcasting Rules

NumPy compares dimensions from right to left.

Two dimensions are compatible if:

They are equal.
One of them is 1.

Otherwise, broadcasting fails.

```python
#Example works
a=np.array([[1],[2],[3]])

b=np.array([10,20,30])

print(a+b)
```

```python
#Example error
a=np.array([1,2])

b=np.array([1,2,3])

print(a+b)
```

Question 17. What is Vectorization?
Easy Explanation

Vectorization means performing operations on an entire array at once instead of using loops.

```python
#Without Vectorization
numbers=[1,2,3,4]

result=[]

for i in numbers:
    result.append(i*2)

print(result)
```

```python
#With Vectorization
arr=np.array([1,2,3,4])

print(arr*2)
```

Question 20. Difference Between View and Copy
View

A view shares memory with the original array.

Changing one affects the other.

A copy creates a completely new array.

```python
arr=np.array([1,2,3])

view=arr.view()

view[0]=100

print(arr)
```

```python
#copy
arr=np.array([1,2,3])

copy=arr.copy()

copy[0]=100

print(arr)
```

21. How do you concatenate multiple arrays?

Concatenation means joining arrays together.

```python
#1D Example
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
```
